```python
import numpy as np
import matplotlib.pyplot as plt

def bundle_to_plot_data(bundle, fit_row=0, use_weighted=False):
    """
    Convierte un bundle tipo H_ a un diccionario estándar para plotting.

    Parámetros
    ----------
    bundle : dict
        Diccionario con llaves como 'Br', 'fit', 'meta'
    fit_row : int
        Qué fila de bundle['fit'] usar
    use_weighted : bool
        Si True usa 'Weighted B(r)', si False usa 'Unweighted B(r)'
    """
    Br = bundle['Br']
    fit = bundle['fit']
    meta = bundle['meta']

    bcol = 'Weighted B(r)' if use_weighted else 'Unweighted B(r)'

    mask = np.array(Br['N pairs']) > 0

    r = (10 ** np.array(Br['log10 r'])[mask]) * meta['PIX'] * meta['PC']
    B = np.array(Br[bcol])[mask]

    d = {
        'r': r,
        'B': B,
        'sig': meta.get('SIG', None),
        'sig2_meta': meta.get('SIG2', None),   # valor proveniente de meta
        's0_meta': meta.get('S0', None),       # valor proveniente de meta
        'pc': meta.get('PC', None),
        'pix': meta.get('PIX', None),
        'line': meta.get('LINE', None),
        'binsize': meta.get('BINSIZE', None),

        # parámetros del ajuste
        'sig2': fit.iloc[fit_row]['sig2'] if 'sig2' in fit.columns else None,
        'r0':   fit.iloc[fit_row]['r0']   if 'r0'   in fit.columns else None,
        'm':    fit.iloc[fit_row]['m']    if 'm'    in fit.columns else None,
        's0':   fit.iloc[fit_row]['s0']   if 's0'   in fit.columns else None,
        'noise':fit.iloc[fit_row]['noise']if 'noise'in fit.columns else None,
    }

    return d
```

```python
import numpy as np

def find_r_where_B_equals_sig2(r, B, sig2):
    """
    Find r where B(r) crosses sig2 using interpolation in log10(r).

    Returns
    -------
    r_cross : float or None
        First crossing location, or None if no crossing exists.
    """
    r = np.asarray(r, dtype=float)
    B = np.asarray(B, dtype=float)

    good = np.isfinite(r) & np.isfinite(B) & (r > 0)
    r = r[good]
    B = B[good]

    if len(r) < 2 or sig2 is None or not np.isfinite(sig2):
        return None

    diff = B - sig2
    idx = np.where(np.diff(np.sign(diff)) != 0)[0]

    if len(idx) == 0:
        return None

    j = idx[0]   # first crossing
    r1, r2 = r[j], r[j + 1]
    d1, d2 = diff[j], diff[j + 1]

    if d2 == d1:
        return None

    logr1, logr2 = np.log10(r1), np.log10(r2)
    logr_cross = logr1 - d1 * (logr2 - logr1) / (d2 - d1)

    return 10**logr_cross
```

```python
import numpy as np
import matplotlib.pyplot as plt

def find_r_where_B_equals_sig2(r, B, sig2):
    """
    Find the first r where B(r) crosses sig2,
    using interpolation in log10(r).
    """
    r = np.asarray(r, dtype=float)
    B = np.asarray(B, dtype=float)

    good = np.isfinite(r) & np.isfinite(B) & (r > 0)
    r = r[good]
    B = B[good]

    if len(r) < 2 or sig2 is None or not np.isfinite(sig2):
        return None

    diff = B - sig2
    idx = np.where(np.diff(np.sign(diff)) != 0)[0]

    if len(idx) == 0:
        return None

    j = idx[0]
    r1, r2 = r[j], r[j + 1]
    d1, d2 = diff[j], diff[j + 1]

    if d2 == d1:
        return None

    logr1, logr2 = np.log10(r1), np.log10(r2)
    logr_cross = logr1 - d1 * (logr2 - logr1) / (d2 - d1)

    return 10**logr_cross


fig, ax = plt.subplots(figsize=(7, 6))

ref = 'Bin 0'
sig2_ref = data[ref]['sig2_meta']

for i, name in enumerate(data_in):
    if name == ref:
        continue

    r_num = np.asarray(data[name]['r'])
    B_num = np.asarray(data[name]['B'])
    r_ref = np.asarray(data[ref]['r'])
    B_ref = np.asarray(data[ref]['B'])

    # ratio only on overlap
    mask = (
        np.isfinite(r_num) & np.isfinite(B_num) &
        (r_num >= r_ref.min()) & (r_num <= r_ref.max())
    )

    r_use = r_num[mask]
    B_use = B_num[mask]
    B_ref_interp = np.interp(r_use, r_ref, B_ref)

    ratio = B_use / B_ref_interp

    # ratio curve
    ax.plot(
        r_use,
        ratio,
        color=colors[i],
        linewidth=1.5,
        alpha=0.9
    )

    ax.scatter(
        r_use,
        ratio,
        label=f'{name} / {ref}',
        color=colors[i],
        zorder=10,
        alpha=0.7,
        s=20
    )

    # r0 point on ratio curve
    r0 = data[name]['r0']
    if r0 is not None and np.isfinite(r0) and len(r_use) > 1:
        if (r0 >= r_use.min()) and (r0 <= r_use.max()):
            ratio_r0 = np.interp(r0, r_use, ratio)

            ax.scatter(
                r0,
                ratio_r0,
                color=colors[i],
                edgecolor='black',
                s=90,
                zorder=20,
                marker='o'
            )
        else:
            # optional: mark r0 even if point cannot be evaluated on ratio curve
            ax.axvline(
                r0,
                color=colors[i],
                linestyle='--',
                linewidth=1.0,
                alpha=0.6
            )

    # horizontal SIG2 ratio line
    sig2_num = data[name]['sig2_meta']
    if (
        sig2_num is not None and sig2_ref is not None
        and np.isfinite(sig2_num) and np.isfinite(sig2_ref)
        and sig2_ref != 0
    ):
        sig2_ratio = sig2_num / sig2_ref

        ax.axhline(
            sig2_ratio,
            color=colors[i],
            linestyle=':',
            linewidth=1.2,
            alpha=0.9
        )

    # vertical line where native B reaches native SIG2
    r_cross = find_r_where_B_equals_sig2(r_num, B_num, sig2_num)
    if r_cross is not None:
        ax.axvline(
            r_cross,
            color=colors[i],
            linestyle='-.',
            linewidth=1.2,
            alpha=0.9
        )

ax.set_xscale('log')
# ax.set_yscale('log')
ax.set_xlabel('Separation r [pc]')
ax.set_ylabel(r'$B_{\mathrm{BinX}}(r) / B_{\mathrm{Bin\ 0}}(r)$')
ax.set_title(f"{data[ref]['line']} ratios")
ax.axhline(1.0, color='black', linestyle='--', linewidth=1)
ax.axvline(128, color='black', linestyle='--', linewidth=1)
ax.axvline(1024, color='black', linestyle='-', linewidth=1)

ax.legend(loc='best', fontsize=7)

plt.tight_layout()
plt.show()
```
