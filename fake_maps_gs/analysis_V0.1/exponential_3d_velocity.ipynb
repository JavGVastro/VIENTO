```python
import pyvista as pv
import gstools as gs
import numpy as np
from astropy.io import fits
```

```python
pv.set_jupyter_backend("trame")
#pv.set_jupyter_backend("client")
```

```python
N = 256
dim, spacing = (N, N, N), (1,1,1)
grid = pv.ImageData(dimensions=dim, spacing=spacing)
```

```python
field_synth = 'velocity'
model_gs    = 'exponential'
alpha       = 1.0
len_scale   = [16, 16, 16]
angles      = [0.8, 0.8, 0.8]
seed        = 19880127
```

```python
model = gs.Exponential(dim=3, len_scale=len_scale, angles=angles)
srf = gs.SRF(model, seed=seed)
```

```python
field = srf.mesh
```

```python
srf.mesh(grid, points='points', name='random-field')
```

```python
grid.contour(isosurfaces=6).plot(smooth_shading=True, use_transparency=False, opacity=0.3, cmap='coolwarm')
```

```python
field_3d = np.asarray(grid.point_data['random-field']).reshape(dim, order='F')
```

```python
print(type(field_3d))
print(field_3d.shape)
print(field_3d.min())
print(field_3d.max())
print(field_3d.mean())
print(field_3d.std())
```

```python
header = fits.Header()

header["MODEL"] = model_gs
header["SEED"]  = seed
header["M3D"]   = alpha

header["LSCALE_X"] = len_scale[0]
header["LSCALE_Y"] = len_scale[1]
header["LSCALE_Z"] = len_scale[2]

header["ANGLE_X"] = angles[0]
header["ANGLE_Y"] = angles[1]
header["ANGLE_Z"] = angles[2]

header["DX"] = spacing[0]
header["DY"] = spacing[1]
header["DZ"] = spacing[2]
```

```python
name = model_gs + "_3d_" + field_synth 
print(name)
```
