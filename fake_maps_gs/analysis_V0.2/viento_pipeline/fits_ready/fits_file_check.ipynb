```python
from astropy.io import fits
from astropy.table import Table

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
model = 'gaussian'
r0 = '13'

entries = {
    "sigE_0": "gstools_finite_sigE0_" + model + "_r" +r0 + "_N256_bin0",
    "sigE_1": "gstools_finite_sigE1_" + model + "_r" +r0 + "_N256_bin0",
    "sigE_2": "gstools_finite_sigE2_" + model + "_r" +r0 + "_N256_bin0",
}
```

```python
data = {}

for alias, name in entries.items():

    with fits.open(name + ".fits") as hdul:

        data[alias] = {
            "name": name,
            "sb": hdul[1].data.copy(),
            "vv": hdul[2].data.copy(),
            "hdr": hdul[0].header.copy(),
        }
```

```python
data['sigE_0']['vv']
```

```python
# ---------------------------------------------------------
# Grayscale palette
# ---------------------------------------------------------

n_samples = len(data)

colors = sns.color_palette(
    "Greys",
    n_colors=n_samples + 2,
)[2:]


# ---------------------------------------------------------
# Figure
# ---------------------------------------------------------

fig, (ax, axx) = plt.subplots(
    1,
    2,
    sharey=False,
    figsize=(12, 6),
)


# ---------------------------------------------------------
# Histograms
# ---------------------------------------------------------

for (alias, entry), color in zip(data.items(), colors):

    # Flatten arrays
    sb = entry["sb"].ravel()
    vv = entry["vv"].ravel()

    # Convert FITS big-endian arrays to native byte order
    sb = np.asarray(sb, dtype=np.float64)
    vv = np.asarray(vv, dtype=np.float64)

    # -----------------------------------------------------
    # Statistics
    # -----------------------------------------------------

    sb_std = sb.std()
    sb_var = sb.var()

    vv_std = vv.std()
    vv_var = vv.var()

    # Labels for legends
    sb_label = (
        f"{alias}\n"
        f"std={sb_std:.3f}, var={sb_var:.3f}"
    )

    vv_label = (
        f"{alias}\n"
        f"std={vv_std:.3f}, var={vv_var:.3f}"
    )


    # -----------------------------------------------------
    # Surface brightness
    # -----------------------------------------------------

    sns.histplot(
        x=sb,
        bins=5000,
        element="step",
        fill=False,
        color=color,
        label=sb_label,
        ax=ax,
    )


    # -----------------------------------------------------
    # Velocity
    # -----------------------------------------------------

    sns.histplot(
        x=vv,
        bins=150,
        element="step",
        fill=False,
        color=color,
        label=vv_label,
        ax=axx,
    )


# ---------------------------------------------------------
# Y limits
# ---------------------------------------------------------

ax.set_ylim(0, 1000)
axx.set_ylim(0, 2000)


# ---------------------------------------------------------
# Axis scales
# ---------------------------------------------------------

ax.set_xscale("log")
# ax.set_yscale("log")


# ---------------------------------------------------------
# Labels
# ---------------------------------------------------------

ax.set_xlabel("Surface brightness")
ax.set_ylabel("Count")

axx.set_xlabel("Velocity")
axx.set_ylabel("Count")


# ---------------------------------------------------------
# Legends
# ---------------------------------------------------------

ax.legend(
    title="Surface brightness",
    fontsize=9,
)

axx.legend(
    title="Velocity",
    fontsize=9,
)


# ---------------------------------------------------------
# Figure title
# ---------------------------------------------------------

fig.suptitle(model)

plt.tight_layout()


# ---------------------------------------------------------
# Save
# ---------------------------------------------------------

fig.savefig(
    "pdf_" + model + ".jpg",
    dpi=300,
    bbox_inches="tight",
)

plt.show()
```
