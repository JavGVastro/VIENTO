```python
from astroquery.skyview import SkyView
from astropy.coordinates import SkyCoord
import astropy.units as u

center = SkyCoord(
    ra=84.675 * u.deg,
    dec=-69.100 * u.deg,
    frame="icrs"
)

images = SkyView.get_images(
    position=center,
    survey=["DSS2 Red"],
    width=1.5 * u.deg,
    height=1.5 * u.deg,
    pixels=2000,
)

images[0][0].writeto(
    "30Dor_DSS2_Red_1p5deg.fits",
    overwrite=True
)
```
