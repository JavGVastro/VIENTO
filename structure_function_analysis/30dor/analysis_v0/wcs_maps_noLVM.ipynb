# Load data

```python
def load_structure_function_dataset(info):
    """
    Load one FITS file containing structure-function data.

    This version:
    - reads the structure-function table from HDU='TABLE'
    - reads MOM0, MOM1, MOM2, MASK if they exist
    - reads HDU 1 as a table if HDU 1 is a BinTableHDU
    - stores everything inside the dataset dictionary
    """

    filename = info["filename"] + ".fits"

    with fits.open(filename) as hdul:

        # Copy primary header
        hdr = hdul[0].header.copy()

        # Get list of HDU names
        hdu_names = [hdu.name for hdu in hdul]

        # Store HDU names for later inspection
        info["hdu_names"] = hdu_names

        # -------------------------------------------------
        # Read HDU 1 only if it is a table
        # -------------------------------------------------
        if len(hdul) > 1 and isinstance(hdul[1], BinTableHDU):

            data_hdu1 = Table(hdul[1].data)
            frame_hdu1 = data_hdu1.to_pandas()

            info["data_hdu1"] = data_hdu1
            info["frame_hdu1"] = frame_hdu1
            info["hdu1_name"] = hdul[1].name

        # -------------------------------------------------
        # Read image HDUs if they exist
        # -------------------------------------------------
        if "MOM0" in hdu_names:
            info["mom0"] = hdul["MOM0"].data.copy()

        if "MOM1" in hdu_names:
            info["mom1"] = hdul["MOM1"].data.copy()

        if "MOM2" in hdu_names:
            info["mom2"] = hdul["MOM2"].data.copy()

        if "MASK" in hdu_names:
            info["mask_image"] = hdul["MASK"].data.copy()

    # -------------------------------------------------
    # Read structure-function table
    # -------------------------------------------------
    data = Table.read(filename, hdu="TABLE")
    frame = data.to_pandas()

    # Mask valid structure-function bins
    if "N pairs" in frame.columns:
        sf_mask = np.array(frame["N pairs"]) > 0
    else:
        sf_mask = np.ones(len(frame), dtype=bool)

    # Extract B(r)
    B = np.array(frame["Unweighted B(r)"])[sf_mask]

    # Extract r
    r = 10 ** np.array(frame["log10 r"])[sf_mask]

    # Optional physical scaling
    if info.get("multiply_r_by_pix_pc", False):

        required_keys = ["PIX", "PC"]
        missing_keys = [k for k in required_keys if k not in hdr]

        if missing_keys:
            raise KeyError(
                f"{info['filename']} requires PIX and PC scaling, "
                f"but these header keys are missing: {missing_keys}"
            )

        pix = hdr["PIX"]
        pc = hdr["PC"]

        r = r * pix * pc

        info["pix"] = pix
        info["pc"] = pc

    # Optional header values
    if "SIG2" in hdr:
        info["sig2"] = hdr["SIG2"]

    # Store table and processed structure-function arrays
    info["table"] = data
    info["frame"] = frame
    info["header"] = hdr
    info["B"] = B
    info["r"] = r
    info["sf_mask"] = sf_mask

    return info
```

```python
for key, info in datasets.items():
    datasets[key] = load_structure_function_dataset(info)
```

```python
from astropy.coordinates import SkyCoord
from astropy.wcs import WCS
import astropy.units as u
c0 = SkyCoord.from_name("Tarantula")
c0

ra_deg = c0.ra.deg
dec_deg = c0.dec.deg
```

```python


# ============================================================
# 1. USER INPUTS
# ============================================================

# RA boundaries (degrees)
RA_MIN = 84.000000     # 05h36m
RA_MAX = 85.250000     # 05h41m

# Dec boundaries (degrees)
DEC_MIN = -69.333333   # -69d20m
DEC_MAX = -68.916667   # -68d55m

# Additional padding in pixels
PADDING = 10

# ============================================================
# 2. LOAD THE ESO IMAGE
# ============================================================

Image.MAX_IMAGE_PIXELS = 130_000_000

image = Image.open("eso1033a.jpg").convert("RGB")

nx, ny = image.size

# Reference dimensions from AstroPix
nx_ref, ny_ref = 9246, 12360

fx = nx / nx_ref
fy = ny / ny_ref

# ============================================================
# 3. RECONSTRUCT THE WCS
# ============================================================

wcs = WCS(naxis=2)

wcs.wcs.ctype = ["RA---TAN", "DEC--TAN"]
wcs.wcs.cunit = ["deg", "deg"]
wcs.wcs.radesys = "ICRS"

wcs.wcs.crval = [84.658040, -69.094994]

wcs.wcs.crpix = [
    (4486.46132767 - 0.5) * fx + 0.5,
    (7693.06622957 - 0.5) * fy + 0.5
]

sx = -9.470457e-05 / fx
sy = +9.470457e-05 / fy

theta = np.deg2rad(0.42914537333)

wcs.wcs.cd = [
    [sx * np.cos(theta), -sy * np.sin(theta)],
    [sx * np.sin(theta),  sy * np.cos(theta)]
]

# ============================================================
# 4. CONVERT SKY BOUNDARIES TO PIXEL BOUNDARIES
# ============================================================

# Sample all four edges of the requested RA/Dec rectangle.
# This accounts for the projection and image orientation.

t = np.linspace(0, 1, 100)

ra_edges = np.concatenate([
    np.full_like(t, RA_MIN),
    np.full_like(t, RA_MAX),
    RA_MIN + (RA_MAX - RA_MIN) * t,
    RA_MIN + (RA_MAX - RA_MIN) * t
])

dec_edges = np.concatenate([
    DEC_MIN + (DEC_MAX - DEC_MIN) * t,
    DEC_MIN + (DEC_MAX - DEC_MIN) * t,
    np.full_like(t, DEC_MIN),
    np.full_like(t, DEC_MAX)
])

x_edges, y_edges = wcs.world_to_pixel_values(
    ra_edges, dec_edges
)

# Bounding rectangle in pixel coordinates
x0 = max(0, int(np.floor(x_edges.min())) - PADDING)
x1 = min(nx, int(np.ceil(x_edges.max())) + PADDING + 1)

y0 = max(0, int(np.floor(y_edges.min())) - PADDING)
y1 = min(ny, int(np.ceil(y_edges.max())) + PADDING + 1)

# ============================================================
# 5. CROP THE IMAGE AND UPDATE WCS
# ============================================================

# PIL uses a top-left origin.
# The reconstructed WCS uses the FITS bottom-left convention.

image_crop = image.crop((
    x0,
    ny - y1,
    x1,
    ny - y0
))

rgb_crop = np.asarray(image_crop)[::-1, :, :]

# Update WCS for the cropped image
wcs_crop = wcs.slice((
    slice(y0, y1),
    slice(x0, x1)
))

print("Original image:", (ny, nx))
print("Cropped image:", rgb_crop.shape)

# ============================================================
# 6. PLOT THE CROPPED IMAGE
# ============================================================

fig = plt.figure(figsize=(10, 9))

ax = fig.add_subplot(
    111,
    projection=wcs_crop
)

ax.imshow(
    rgb_crop,
    origin="lower"
)

# ============================================================
# 7. OVERLAY OBSERVATIONAL DATA
# ============================================================

# FLAMES
ax.scatter(
    datasets["FLA"]["frame_hdu1"]["RAdeg"],
    datasets["FLA"]["frame_hdu1"]["DEdeg"],
    s=4,
    alpha=0.75,
    label="FLAMES",
    marker="x",
    color="lime",
    transform=ax.get_transform("world"),
    zorder=3
)

# MUSE
ax.scatter(
    df1["RAdeg"],
    df1["DEdeg"],
    s=4,
    marker="s",
    alpha=0.01,
    label="MUSE",
    transform=ax.get_transform("world"),
    zorder=3
)

# SAMFP
ax.scatter(
    df_fp["RAdeg"],
    df_fp["DEdeg"],
    s=4,
    marker="s",
    alpha=0.01,
    label="SAMFP",
    transform=ax.get_transform("world"),
    zorder=3
)

# ============================================================
# 8. AXES AND FORMATTING
# ============================================================

ax.coords.grid(
    color="white",
    alpha=0.4,
    linestyle=":"
)

# Sexagesimal formatting
ax.coords[0].set_major_formatter("hh:mm")
ax.coords[1].set_major_formatter("dd:mm")

ax.coords[0].set_ticks(spacing=15 * u.arcmin)
ax.coords[1].set_ticks(spacing=5 * u.arcmin)

ax.set_xlabel("Right Ascension (J2000)")
ax.set_ylabel("Declination (J2000)")

# R136 position
ax.scatter(
    c0.ra.deg,
    c0.dec.deg,
    transform=ax.get_transform("world"),
    color="white",
#    edgecolor="white",  
    s=80,
    marker="*",
    zorder=5
)

#ax.annotate(
#    "R136",
#    xy=(c0.ra.deg, c0.dec.deg),
#    xycoords=ax.get_transform("world"),
#    xytext=(8, 8),
#    textcoords="offset points",
#    color="white",
#    fontsize=14,
#    ha="left",
#    va="bottom"
#)

ax.legend(
    loc="upper right",
    markerscale=5,
    fontsize=12,
    framealpha=0.8
)




plt.tight_layout()
plt.show()
```

```python
from scipy.spatial import ConvexHull

def plot_fov_hull(ax, ra, dec, **plot_kwargs):
    points = np.column_stack([ra, dec])
    hull = ConvexHull(points)
    hull_pts = points[hull.vertices]
    hull_pts = np.vstack([hull_pts, hull_pts[0]])

    ax.plot(
        hull_pts[:, 0],
        hull_pts[:, 1],
        transform=ax.get_transform("world"),
        **plot_kwargs
    )
```

```python

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
import astropy.units as u

from PIL import Image
from astropy.wcs import WCS

# ============================================================
# 1. USER INPUTS
# ============================================================

# RA boundaries (degrees)
RA_MIN = 84.000000     # 05h36m
RA_MAX = 85.250000     # 05h41m

# Dec boundaries (degrees)
DEC_MIN = -69.333333   # -69d20m
DEC_MAX = -68.916667   # -68d55m

# Additional padding in pixels
PADDING = 10

# Physical scale bar
DISTANCE_KPC = 50.0      # Distance to 30 Doradus (kpc)
SCALE_BAR_PC = 50.0      # Scale bar length (pc)

# Scale bar position (fraction of image width/height)
SCALE_BAR_X = 0.08       # From left
SCALE_BAR_Y = 0.08       # From bottom

# Scale bar appearance
SCALE_BAR_COLOR = "white"
SCALE_BAR_LINEWIDTH = 3
SCALE_BAR_FONTSIZE = 14

# ============================================================
# 2. LOAD THE ESO IMAGE
# ============================================================

Image.MAX_IMAGE_PIXELS = 130_000_000

image = Image.open("eso1033a.jpg").convert("RGB")

nx, ny = image.size

# Reference dimensions from AstroPix
nx_ref, ny_ref = 9246, 12360

fx = nx / nx_ref
fy = ny / ny_ref

# ============================================================
# 3. RECONSTRUCT THE WCS
# ============================================================

wcs = WCS(naxis=2)

wcs.wcs.ctype = ["RA---TAN", "DEC--TAN"]
wcs.wcs.cunit = ["deg", "deg"]
wcs.wcs.radesys = "ICRS"

wcs.wcs.crval = [84.658040, -69.094994]

wcs.wcs.crpix = [
    (4486.46132767 - 0.5) * fx + 0.5,
    (7693.06622957 - 0.5) * fy + 0.5
]

sx = -9.470457e-05 / fx
sy = +9.470457e-05 / fy

theta = np.deg2rad(0.42914537333)

wcs.wcs.cd = [
    [sx * np.cos(theta), -sy * np.sin(theta)],
    [sx * np.sin(theta),  sy * np.cos(theta)]
]

# ============================================================
# 4. CONVERT SKY BOUNDARIES TO PIXEL BOUNDARIES
# ============================================================

t = np.linspace(0, 1, 100)

ra_edges = np.concatenate([
    np.full_like(t, RA_MIN),
    np.full_like(t, RA_MAX),
    RA_MIN + (RA_MAX - RA_MIN) * t,
    RA_MIN + (RA_MAX - RA_MIN) * t
])

dec_edges = np.concatenate([
    DEC_MIN + (DEC_MAX - DEC_MIN) * t,
    DEC_MIN + (DEC_MAX - DEC_MIN) * t,
    np.full_like(t, DEC_MIN),
    np.full_like(t, DEC_MAX)
])

x_edges, y_edges = wcs.world_to_pixel_values(
    ra_edges, dec_edges
)

x0 = max(0, int(np.floor(x_edges.min())) - PADDING)
x1 = min(nx, int(np.ceil(x_edges.max())) + PADDING + 1)

y0 = max(0, int(np.floor(y_edges.min())) - PADDING)
y1 = min(ny, int(np.ceil(y_edges.max())) + PADDING + 1)

# ============================================================
# 5. CROP THE IMAGE AND UPDATE WCS
# ============================================================

image_crop = image.crop((
    x0,
    ny - y1,
    x1,
    ny - y0
))

rgb_crop = np.asarray(image_crop)[::-1, :, :]

wcs_crop = wcs.slice((
    slice(y0, y1),
    slice(x0, x1)
))

print("Original image:", (ny, nx))
print("Cropped image:", rgb_crop.shape)

# ============================================================
# 6. PLOT THE CROPPED IMAGE
# ============================================================

fig = plt.figure(figsize=(10, 9))

ax = fig.add_subplot(
    111,
    projection=wcs_crop
)

ax.imshow(
    rgb_crop,
    origin="lower"
)

# ============================================================
# 7. OVERLAY OBSERVATIONAL DATA
# ============================================================

# FLAMES
ax.scatter(
    datasets["FLA"]["frame_hdu1"]["RAdeg"],
    datasets["FLA"]["frame_hdu1"]["DEdeg"],
    s=4,
    alpha=0.75,
    label="FLAMES",
    marker="o",
    color="lime",
    transform=ax.get_transform("world"),
    zorder=3
)

# MUSE
plot_fov_hull(
    ax,
    df1["RAdeg"].values,
    df1["DEdeg"].values,
    color="white",
    linestyle="-.",
    linewidth=2,
    label="MUSE",
    zorder=4
)

# SAMFP
plot_fov_hull(
    ax,
    df_fp["RAdeg"].values,
    df_fp["DEdeg"].values,
    color="white",
    linewidth=2,
    label="SAMFP",
    zorder=4
)

# ============================================================
# 8. AXES AND FORMATTING
# ============================================================

ax.coords.grid(
    color="white",
    alpha=0.4,
    linestyle=":"
)

# Sexagesimal formatting
ax.coords[0].set_major_formatter("hh:mm")
ax.coords[1].set_major_formatter("dd:mm")

ax.coords[0].set_ticks(spacing=15 * u.arcmin)
ax.coords[1].set_ticks(spacing=5 * u.arcmin)

ax.set_xlabel("Right Ascension (J2000)")
ax.set_ylabel("Declination (J2000)")

# R136 position
ax.scatter(
    c0.ra.deg,
    c0.dec.deg,
    transform=ax.get_transform("world"),
    color="white",
    s=80,
    marker="*",
    zorder=5,
    label="R136"
)

ax.legend(
    loc="upper right",
    markerscale=2,
    fontsize=12,
    framealpha=0.8
)

# ============================================================
# 9. PHYSICAL SCALE BAR
# ============================================================

# Convert physical size to angular separation
distance = DISTANCE_KPC * u.kpc
scale_length = SCALE_BAR_PC * u.pc

angular_size = (
    np.arctan((scale_length / distance).decompose().value)
    * u.rad
)

# Get the displayed image limits
xlim = ax.get_xlim()
ylim = ax.get_ylim()

xmin, xmax = sorted(xlim)
ymin, ymax = sorted(ylim)

# Position in image pixel coordinates
x_bar = xmin + SCALE_BAR_X * (xmax - xmin)
y_bar = ymin + SCALE_BAR_Y * (ymax - ymin)

# Calculate local angular pixel scale using the WCS
coord1 = wcs_crop.pixel_to_world(x_bar, y_bar)
coord2 = wcs_crop.pixel_to_world(x_bar + 1, y_bar)

pixel_scale = coord1.separation(coord2)

# Convert physical scale into pixel length
bar_pixels = (
    angular_size.to_value(u.arcsec)
    / pixel_scale.to_value(u.arcsec)
)

# Draw the horizontal scale bar
ax.plot(
    [x_bar, x_bar + bar_pixels],
    [y_bar, y_bar],
    color=SCALE_BAR_COLOR,
    linewidth=SCALE_BAR_LINEWIDTH,
    solid_capstyle="butt",
    transform=ax.transData,
    path_effects=[
        pe.Stroke(
            linewidth=SCALE_BAR_LINEWIDTH + 2,
            foreground="black"
        ),
        pe.Normal()
    ],
    zorder=10
)

# Add the physical scale label
ax.annotate(
    f"{SCALE_BAR_PC:g} pc",
    xy=(x_bar + bar_pixels / 2, y_bar),
    xycoords="data",
    xytext=(0, 7),
    textcoords="offset points",
    color=SCALE_BAR_COLOR,
    fontsize=SCALE_BAR_FONTSIZE,
    ha="center",
    va="bottom",
    path_effects=[
        pe.withStroke(
            linewidth=3,
            foreground="black"
        )
    ],
    zorder=11
)

print(f"Distance: {DISTANCE_KPC:g} kpc")
print(f"Scale bar: {SCALE_BAR_PC:g} pc")
print(f"Angular size: {angular_size.to(u.arcmin):.3f}")
print(f"Scale bar length: {bar_pixels:.1f} pixels")

# ============================================================
# 10. EXPORT FIGURE
# ============================================================

plt.tight_layout()

fig.savefig(
    "30Dor_instruments_FoV.pdf",
    format="pdf",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
