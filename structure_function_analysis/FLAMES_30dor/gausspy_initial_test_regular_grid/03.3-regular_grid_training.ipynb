```python
import gausspy.gp as gp

FILENAME_TRAIN = "flames30dor_training_data_composition_03.pkl"

snr_thresh = 5.0
alpha1_initial = 1.0
alpha2_initial = 2.0

g = gp.GaussianDecomposer()
g.load_training_data(FILENAME_TRAIN)

g.set("phase", "two")
g.set("SNR_thresh", [snr_thresh, snr_thresh])

g.train(alpha1_initial=alpha1_initial, alpha2_initial=alpha2_initial)
```

```python
import pickle

with open("gausspy_training_FLAMES_30Dor_two_phase_03.pkl", "wb") as f:
    pickle.dump(g.p, f)
```
