```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

manifest = pd.read_csv("data_comparison/manifest.csv")
data_folder = Path(".")

active = manifest[manifest["enabled"] == 1]

plt.figure(figsize=(9, 6))

for _, row in active.iterrows():
    file_path = data_folder / row["filename"]
    
    if not file_path.exists():
        print(f"Warning: {row['filename']} not found, skipping.")
        continue
    
    df = pd.read_csv(file_path)
    x_col = df.columns[0]
    y_col = df.columns[1]
    
    plt.plot(df[x_col], df[y_col], label=row["label"])

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Selected CSV Plots")
plt.grid(True)
plt.legend()
plt.show()
```
